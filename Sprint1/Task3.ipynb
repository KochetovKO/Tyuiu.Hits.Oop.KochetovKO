```

public class Customer
{
    public int CustomerID {get; set;}
    public string Name {get; set;}
    public string Email {get; set;}
    
    public Customer(int customerID, string name, string email)
    {
        CustomerID = customerID;
        Name = name;
        Email = email;
    }
    public void InviteFriend(Customer Friend)
    {
        Console.WriteLine($"покупатель {Name} пригласил друга {Friend.Name}");
        Console.WriteLine($"покупатель {Name} получает скидку 10%, реферал {Friend.Name} получает скидку 7%");
    }
    
    public virtual string GetFullName()
    {
        return Name;
    }

    public virtual void UpdateEmail(string newEmail)
    {
        Console.WriteLine($"{Email} клиента {Name} изменен на {newEmail}");
        Email = newEmail;
    }

    public virtual void ViewProfile()
    {
        Console.WriteLine($"Профиль Клиента ID : {CustomerID}");
        Console.WriteLine($"Имя клиента : {Name}");
        Console.WriteLine($"Почта клиента : {Email}");
    }


}
public class VIPCustomer : Customer
{
    public int LoyaltyPoints{get; set;}

    public VIPCustomer(int customerID, string name, string email, int loyaltyPoints) : base(customerID, name, email)
    {
        LoyaltyPoints = loyaltyPoints;
    }
    
    public override void ViewProfile()
    {
        Console.WriteLine("------VIP клиент------");
        base.ViewProfile();
        Console.WriteLine($"Баланс лояльности : {LoyaltyPoints}");
    }
}
public class RegularCustomer : Customer
{
    public RegularCustomer(int customerID, string name, string email) : base(customerID, name, email){}
    
    public DateTime RegistrationDate {get; set;}

   public override void UpdateEmail(string newEmail)
    {
        base.UpdateEmail(newEmail);
        Console.WriteLine($"Дата и время обновления : {DateTime.Now}");
    }    
}
public class GroupCustomer : Customer
{
    public GroupCustomer(int customerID, string groupname, string email) : base(customerID, groupname, email)
    {
        GroupName = groupname;
    }

    public string GroupName {get; set;}

    public override string GetFullName()
    {
        return GroupName;
    }
}
// обычный клиент
RegularCustomer myRegularCustomer = new RegularCustomer(1, "Andrey", "andrey@mail.ru");
myRegularCustomer.UpdateEmail("andrey222@mail.ru");
myRegularCustomer.ViewProfile();


// VIP клиент
Console.WriteLine("");
VIPCustomer myVIPCustomer = new VIPCustomer(2, "Pavel", "pavel@mail.ru", 1200);
myVIPCustomer.ViewProfile();
myVIPCustomer.InviteFriend(myRegularCustomer);

//Компания
Console.WriteLine("");
GroupCustomer myGroupCustomer = new GroupCustomer(3, "ООО 'Строительный двор'"," stroitDVOR@mail.ru");
myGroupCustomer.ViewProfile();

```
